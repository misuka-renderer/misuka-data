```python
import misuka as mi
mi.set_variant('cuda_acoustic', 'llvm_acoustic', 'scalar_acoustic')
import pyfar as pf
import matplotlib.pyplot as plt
import numpy as np
```

```python
sampling_rate = 1000.0
```

```python
from pathlib import Path

for p in Path('.').glob('*_primal_ref.exr'):
    b = p.stem[:-len('_primal_ref')]
    A = np.array(mi.TensorXf(mi.Bitmap(str(p))))
    fpath = Path(f'{b}_fwd_ref.exr')
    B = np.array(mi.TensorXf(mi.Bitmap(str(fpath)))) if fpath.exists() else None
    sp = pf.Signal(A.T, sampling_rate=sampling_rate)
    sf = pf.Signal(B.T, sampling_rate=sampling_rate) if B is not None else None
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(2,1,sharex=True,figsize=(8,4))
    pf.plot.time(sp, dB=True, ax=ax[0], unit='samples')
    if sf is not None: pf.plot.time(sf, dB=False, ax=ax[1], unit='samples')
    else: ax[1].text(.5,.5,'missing',ha='left',va='center')
    fig.suptitle(b)
    ax[0].set_title('primal rendering')
    ax[1].set_title('forward finite differences')
    plt.tight_layout()
    plt.show()
```
